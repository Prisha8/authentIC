```python
# !pip install -q transformers accelerate timm pillow torch

import cv2, numpy as np, matplotlib.pyplot as plt, torch
from PIL import Image
from transformers import TrOCRProcessor, VisionEncoderDecoderModel

IMG_PATH   = "/kaggle/input/ic-image/ic.jpg"
SAVE_DEBUG = "/kaggle/working/ic_mser_trocr.jpg"

# ------- 1) Preprocess (contrast + light sharpen) -------
def preprocess_ic(bgr):
    g = cv2.cvtColor(bgr, cv2.COLOR_BGR2GRAY)
    # CLAHE for faint engravings
    clahe = cv2.createCLAHE(clipLimit=2.5, tileGridSize=(8,8))
    g = clahe.apply(g)
    # gentle unsharp
    blur = cv2.GaussianBlur(g, (0,0), 1.0)
    g2 = cv2.addWeighted(g, 1.8, blur, -0.8, 0)
    return g2

# ------- 2) Deskew using Hough (align long edge ~ horizontal) -------
def deskew(gray):
    gb = cv2.GaussianBlur(gray, (3,3), 0)
    edges = cv2.Canny(gb, 50, 150)
    lines = cv2.HoughLines(edges, 1, np.pi/180, 120)
    angle = 0.0
    if lines is not None and len(lines) > 0:
        angs = []
        for rho, theta in lines[:,0,:]:
            a = (theta*180/np.pi) - 90
            if a > 90: a -= 180
            if a < -90: a += 180
            angs.append(a)
        angle = float(np.median(angs))
    h, w = gray.shape[:2]
    M = cv2.getRotationMatrix2D((w/2, h/2), angle, 1.0)
    rot = cv2.warpAffine(gray, M, (w, h), flags=cv2.INTER_CUBIC, borderMode=cv2.BORDER_REPLICATE)
    return rot, angle

# ------- 3) MSER proposals + merge (NMS) -------
def mser_boxes(gray):
    boxes = []
    for invert in (False, True):
        img = 255 - gray if invert else gray
        mser = cv2.MSER_create(5)
        mser.setMinArea(60)
        mser.setMaxArea(8000)
        regions, _ = mser.detectRegions(img)
        for r in regions:
            x, y, w, h = cv2.boundingRect(r.reshape(-1, 1, 2))
            ar = w / max(h, 1)
            if w*h < 80 or w < 8 or h < 6:
                continue
            if ar < 0.2 or ar > 20:
                continue
            pad = 2
            boxes.append([x - pad, y - pad, x + w + pad, y + h + pad])
    if not boxes: 
        return []
    boxes = np.array(boxes, dtype=np.float32)

    # Non-maximum suppression
    def nms(bbs, iou_thr=0.3):
        if len(bbs)==0: return []
        x1,y1,x2,y2 = bbs[:,0],bbs[:,1],bbs[:,2],bbs[:,3]
        areas = (x2-x1+1)*(y2-y1+1)
        order = np.argsort((x2-x1)*(y2-y1))  # small to large
        keep=[]
        while order.size>0:
            i = order[-1]
            keep.append(i)
            xx1 = np.maximum(x1[i], x1[order[:-1]])
            yy1 = np.maximum(y1[i], y1[order[:-1]])
            xx2 = np.minimum(x2[i], x2[order[:-1]])
            yy2 = np.minimum(y2[i], y2[order[:-1]])
            w = np.maximum(0, xx2-xx1+1); h=np.maximum(0, yy2-yy1+1)
            inter = w*h
            iou = inter / (areas[i] + areas[order[:-1]] - inter + 1e-6)
            order = order[np.where(iou <= iou_thr)[0]]
        return bbs[keep]

    boxes = nms(boxes, 0.3)

    # Merge nearby boxes into lines (simple y-overlap grouping)
    boxes = boxes.astype(int).tolist()
    boxes.sort(key=lambda b:(b[1], b[0]))
    merged=[]
    for b in boxes:
        if not merged:
            merged.append(b)
            continue
        x1,y1,x2,y2 = b
        mx1,my1,mx2,my2 = merged[-1]
        # if vertical overlap significant and horizontal distance small → merge
        if min(y2,my2) - max(y1,my1) > 0.4*min(y2-y1, my2-my1) and x1 - mx2 < 40:
            merged[-1] = [min(mx1,x1), min(my1,y1), max(mx2,x2), max(my2,y2)]
        else:
            merged.append(b)
    return merged

# ------- 4) TrOCR setup (printed) -------
model_id  = "microsoft/trocr-large-printed"
processor = TrOCRProcessor.from_pretrained(model_id)
model     = VisionEncoderDecoderModel.from_pretrained(model_id)
device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device).eval()

def trocr_read(gray_crop):
    # normalize + upscale for recognizer
    crop = gray_crop
    crop = cv2.resize(crop, None, fx=2.0, fy=2.0, interpolation=cv2.INTER_CUBIC)
    # small de-noise
    crop = cv2.fastNlMeansDenoising(crop, h=7)
    rgb  = cv2.cvtColor(crop, cv2.COLOR_GRAY2RGB)
    pil  = Image.fromarray(rgb)
    ins  = processor(images=pil, return_tensors="pt").to(device)
    with torch.no_grad():
        ids = model.generate(**ins, max_length=64, num_beams=6, early_stopping=True)
    return processor.batch_decode(ids, skip_special_tokens=True)[0].strip()

# ------- 5) Run pipeline -------
bgr = cv2.imread(IMG_PATH)
if bgr is None: raise FileNotFoundError(IMG_PATH)
g0  = preprocess_ic(bgr)
g, ang = deskew(g0)

# Prefer center band crop to avoid legs/background
h, w = g.shape[:2]
band = g[int(h*0.25):int(h*0.75), :]   # middle 50% in height
boxes = mser_boxes(band)

print(f"Deskew≈{ang:.1f}°, boxes found: {len(boxes)}")
if not boxes:
    print("No MSER boxes—try increasing middle band (e.g., 0.2–0.8) or lowering filters.")
    
# read each merged box with TrOCR
reads=[]
vis = cv2.cvtColor(band, cv2.COLOR_GRAY2BGR)
for i,(x1,y1,x2,y2) in enumerate(boxes):
    x1 = max(0,x1); y1 = max(0,y1); x2 = min(band.shape[1]-1,x2); y2 = min(band.shape[0]-1,y2)
    crop = band[y1:y2, x1:x2]
    # ensure landscape
    if crop.shape[0] > crop.shape[1]:
        crop = cv2.rotate(crop, cv2.ROTATE_90_CLOCKWISE)
    txt = trocr_read(crop)
    reads.append(txt)
    cv2.rectangle(vis, (x1,y1), (x2,y2), (0,255,0), 2)
    cv2.putText(vis, txt[:24], (x1, max(16,y1-6)), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,255,0),2)

# ------- 6) Output -------
print("\n=== MSER + TrOCR reads ===")
for i, t in enumerate(reads):
    print(f"[{i}] {t}")

cv2.imwrite(SAVE_DEBUG, vis)
print("\nSaved debug overlay:", SAVE_DEBUG)

plt.figure(figsize=(12,5))
plt.subplot(1,2,1); plt.title("Deskewed center band"); plt.imshow(band, cmap="gray"); plt.axis("off")
plt.subplot(1,2,2); plt.title("MSER boxes + TrOCR text"); plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB)); plt.axis("off")
plt.show()
```

```python
# === IC serial post-filter (NO single final; keep both lists) ===
import re, csv, json

# 1) normalize & quick noise filter
def _norm(s: str) -> str:
    s = (s or "").upper()
    s = re.sub(r"[^A-Z0-9\-_. /]", "", s)
    s = re.sub(r"\s+", " ", s).strip()
    return s

NOISE = {"QTY","ITEM","TOTAL","PCS","EA","EA.","EA:","QTY:","ITEM:"}

def _is_noise(s: str) -> bool:
    s = _norm(s)
    if not s or s in NOISE:
        return True
    core = re.sub(r"[^A-Z0-9]", "", s)
    if len(core) >= 4 and len(set(core)) == 1:  # RRRRRR / IIIIII
        return True
    if core.isdigit():
        return True
    if len(core) < 3:
        return True
    return False

# 2) pattern scoring (tune prefixes for your vendors)
PN_PREFIX = r"(SN|MM|CD|MC|TLV|TPS|TPA|TPS7|TPS6|LM|INA|OPA|ADS|ADC|DAC|MCP|PIC|DSPIC|ATMEGA|ATTINY|STM32|STM8|SAMD|NCP|MAX|AD|LT|LTC|SI|XC|AP|AMS|IRF|AOZ|FAN|BC|BD|BQ|RT|MP|ISL|HC|HCT|74HC|74HCT|74H|74|74LV|SN74|CD40)"
PN_REGEX   = re.compile(rf"^(?:{PN_PREFIX})[A-Z0-9\-_/\.]{{2,}}$")
GENERIC_PN = re.compile(r"^(?=.*[A-Z])(?=.*[0-9])[A-Z0-9][A-Z0-9\-_/\. ]{3,}$")

LOT_TAIL   = re.compile(r".*\b([EGY]\d|E\d|G\d|Y\d)\b")     # E4, G4, Y1…
GRADE_TAIL = re.compile(r".*\b([A-Z]\d{2}|[A-Z]{1,2}\d)\b") # Q1, M4…
SHORT_LOT  = re.compile(r"^[A-Z0-9]{4,}\s?(E\d|G\d)$")

def _pattern_boost(s: str) -> float:
    u = s.replace(" ","")
    score = 0.0
    if PN_REGEX.match(u):       score += 12.0
    elif GENERIC_PN.match(s):   score += 8.0
    if SHORT_LOT.match(s):      score += 5.0
    if LOT_TAIL.match(s):       score += 3.5
    if GRADE_TAIL.match(s):     score += 2.0
    return score

def _shape_score(s: str) -> float:
    core = re.sub(r"[^A-Z0-9]", "", s.upper())
    has_letter = any(c.isalpha() for c in core)
    has_digit  = any(c.isdigit() for c in core)
    length     = len(core)
    length_term = max(0.0, 12.0 - abs(12 - min(length, 24)))  # sweet spot
    combo       = 6.0 if (has_letter and has_digit) else 0.0
    dash_bonus  = 1.0 if "-" in s else 0.0
    penalty_rep = -4.0 if length >= 6 and len(set(core)) <= max(2, length//4) else 0.0
    return length_term + combo + dash_bonus + penalty_rep

def _score(s: str) -> float:
    return _shape_score(s) + _pattern_boost(s)

# ---- INPUT: `reads` from your previous cell ----
cleaned = []
for t in reads:
    if _is_noise(t): 
        continue
    t2 = _norm(t)
    if _is_noise(t2):
        continue
    cleaned.append((t2, _score(t2)))

# de-dup by alnum core, keep best score
best_by_key = {}
for t, sc in cleaned:
    key = re.sub(r"[^A-Z0-9]", "", t)
    if key not in best_by_key or sc > best_by_key[key][1]:
        best_by_key[key] = (t, sc)

ranked = sorted(best_by_key.values(), key=lambda x: x[1], reverse=True)  # [(text,score), …]

# split into two buckets
primary_parts    = [(t, sc) for t, sc in ranked if PN_REGEX.match(t.replace(" ","")) or GENERIC_PN.match(t)]
companion_codes  = [(t, sc) for t, sc in ranked if (t, sc) not in primary_parts]

# ---- OUTPUTS YOU ASKED FOR ----
print("\n=== Ranked candidates (all) ===")
for i,(t,sc) in enumerate(ranked[:20]):
    print(f"[{i:02d}] {t:20s}  score={sc:.2f}")

print("\n=== Likely PART NUMBERS ===")
for i,(t,sc) in enumerate(primary_parts[:10]):
    print(f"[{i:02d}] {t:20s}  score={sc:.2f}")

print("\n=== Companion codes (lot/grade/extra) ===")
for i,(t,sc) in enumerate(companion_codes[:10]):
    print(f"[{i:02d}] {t:20s}  score={sc:.2f}")

# optional: write CSV + JSON
csv_path  = "/kaggle/working/ic_candidates.csv"
json_path = "/kaggle/working/ic_candidates.json"

with open(csv_path, "w", newline="") as f:
    w = csv.writer(f)
    w.writerow(["rank","text","score"])
    for i,(t,sc) in enumerate(ranked):
        w.writerow([i, t, f"{sc:.2f}"])
    w.writerow([])
    w.writerow(["section","text","score"])
    for t,sc in primary_parts:
        w.writerow(["primary", t, f"{sc:.2f}"])
    for t,sc in companion_codes:
        w.writerow(["companion", t, f"{sc:.2f}"])

with open(json_path, "w") as f:
    json.dump({
        "ranked": [{"text":t,"score":sc} for t,sc in ranked],
        "primary_parts": [{"text":t,"score":sc} for t,sc in primary_parts],
        "companion_codes": [{"text":t,"score":sc} for t,sc in companion_codes],
        "raw_reads": list(reads),
    }, f, indent=2)

print("\nSaved:")
print(" - CSV :", csv_path)
print(" - JSON:", json_path)
```

```python
# --- TI datasheet fetcher: symlink -> HTTP scrape -> Playwright fallback (actually runs) ---

# 1) deps (safe to re-run)
# !pip install -q requests beautifulsoup4 lxml playwright
# !python -m playwright install chromium
# (on Kaggle, Playwright needs --no-sandbox when launching)

import os, re, json, csv, requests
from bs4 import BeautifulSoup
from pathlib import Path

# -------- CONFIG --------
INPUT_PATH = "/kaggle/working/ic_candidates.json"  # or .csv from your first notebook
OUT_DIR    = "/kaggle/working/ti_datasheets"
os.makedirs(OUT_DIR, exist_ok=True)

HEADERS = {
    "User-Agent": "Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 "
                  "(KHTML, like Gecko) Chrome/123.0 Safari/537.36"
}

PACKAGE_SUFFIXES = [
    "Q1","E4","G4","T",
    "N","D","DR","DGK","DBR","PW","PWR","PWT","RHL","RGT","RGR","RTER","RGYR",
    "RTWR","RUT","RUTR","RTE","SOIC","TSSOP","QFN","SOT","SOT23"
]

def _alnum(s: str) -> str:
    return re.sub(r"[^A-Z0-9]", "", (s or "").upper())

def to_ti_base_slug(part: str) -> str:
    u = (part or "").upper().replace(" ", "").replace("_","")
    u = re.sub(r"[-\.]+$", "", u)
    changed = True
    while changed:
        changed = False
        for suf in sorted(PACKAGE_SUFFIXES, key=len, reverse=True):
            if u.endswith(suf) and len(u) > len(suf) + 3:
                u = u[:-len(suf)]
                changed = True
                break
    u = re.sub(r"[-_/\.]+$", "", u)
    return u

def ti_product_url(base: str) -> str:
    return f"https://www.ti.com/product/{base}"

def download_pdf(url: str, out_dir: str = OUT_DIR) -> str:
    fname = url.split("/")[-1].split("?")[0] or "ti_datasheet.pdf"
    path = os.path.join(out_dir, fname)
    with requests.get(url, headers=HEADERS, stream=True, timeout=60, allow_redirects=True) as r:
        r.raise_for_status()
        with open(path, "wb") as f:
            for chunk in r.iter_content(1 << 14):
                if chunk: f.write(chunk)
    return path

# -------- load candidates from your file --------
def load_candidates_from_json(path: str) -> list[str]:
    with open(path, "r") as f:
        data = json.load(f)
    texts = []
    for key in ("primary_parts", "companion_codes", "ranked", "raw_reads"):
        if key in data:
            for item in data[key]:
                if isinstance(item, dict) and "text" in item: texts.append(item["text"])
                elif isinstance(item, str): texts.append(item)
    # de-dup
    seen=set(); uniq=[]
    for t in texts:
        k=_alnum(t)
        if k and k not in seen:
            seen.add(k); uniq.append(t)
    return uniq

def load_candidates_from_csv(path: str) -> list[str]:
    texts=[]
    with open(path, "r", newline="") as f:
        rows = list(csv.reader(f))
    for row in rows:
        if not row: continue
        # ranked rows usually look like: rank,text,score
        if len(row) >= 2 and row[0].isdigit():
            texts.append(row[1])
        elif len(row) == 1 and row[0].strip():
            texts.append(row[0].strip())
    # de-dup
    seen=set(); uniq=[]
    for t in texts:
        k=_alnum(t)
        if k and k not in seen:
            seen.add(k); uniq.append(t)
    return uniq

ext = Path(INPUT_PATH).suffix.lower()
all_texts = load_candidates_from_json(INPUT_PATH) if ext==".json" else load_candidates_from_csv(INPUT_PATH)

# Prefer SN* first, then longer tokens
candidates = sorted(
    {re.sub(r"\s+", " ", t.upper()).strip() for t in all_texts if t.strip()},
    key=lambda s: (0 if s.startswith("SN") else 1, -len(_alnum(s)))
)
print("Candidates:", candidates[:10])

# -------- A) try symlink first --------
def try_symlink(base_slug: str):
    slug = base_slug.lower()
    urls = [
        f"https://www.ti.com/lit/ds/symlink/{slug}.pdf",
        f"https://www.ti.com/lit/gpn/{slug}",  # may redirect to PDF
    ]
    for u in urls:
        try:
            r = requests.get(u, headers=HEADERS, timeout=25, allow_redirects=True)
            if r.status_code == 200:
                ctype = (r.headers.get("content-type","") or "").lower()
                # quick PDF check
                if "pdf" in ctype or ("%PDF" in r.content[:1024].decode("latin-1", "ignore")):
                    # re-download via streamer
                    return download_pdf(r.url if "gpn" in u else u)
        except Exception:
            pass
    return None

# -------- B) product page scrape --------
def find_datasheet_via_product(base_slug: str) -> str | None:
    url = ti_product_url(base_slug)
    try:
        r = requests.get(url, headers=HEADERS, timeout=25)
        if r.status_code != 200:
            return None
        soup = BeautifulSoup(r.text, "lxml")
        # datasheet link anchors or /lit/ds/ links
        for a in soup.find_all("a", href=True):
            href = a["href"]
            txt  = (a.get_text() or "").lower()
            h    = href.lower()
            if h.endswith(".pdf") and ("/lit/ds/" in h or "datasheet" in txt):
                return requests.compat.urljoin(url, href)
        for a in soup.find_all("a", href=True):
            h = a["href"].lower()
            if "/lit/ds/" in h and h.endswith(".pdf"):
                return requests.compat.urljoin(url, a["href"])
    except Exception:
        return None
    return None

manifest = {
    "input_path": INPUT_PATH,
    "ok": False,
    "method": None,
    "picked_query": None,
    "base_slug": None,
    "product_url": None,
    "pdf_url": None,
    "pdf_path": None,
    "tried": []
}

# -------- run attempts (symlink -> http) --------
for q in candidates:
    base = to_ti_base_slug(q)
    manifest["tried"].append({"seen": q, "base": base})
    # symlink
    pdf_path = try_symlink(base)
    if pdf_path:
        manifest.update({
            "ok": True, "method": "symlink",
            "picked_query": q, "base_slug": base,
            "product_url": ti_product_url(base),
            "pdf_url": f"https://www.ti.com/lit/ds/symlink/{base.lower()}.pdf",
            "pdf_path": pdf_path
        })
        break
    # http scrape
    pdf_url = find_datasheet_via_product(base)
    if pdf_url:
        try:
            pdf_path = download_pdf(pdf_url)
            manifest.update({
                "ok": True, "method": "http",
                "picked_query": q, "base_slug": base,
                "product_url": ti_product_url(base),
                "pdf_url": pdf_url, "pdf_path": pdf_path
            })
            break
        except Exception:
            pass

# -------- C) Playwright fallback (THIS EXECUTES) --------
if not manifest["ok"]:
    try:
        from playwright.sync_api import sync_playwright
        def playwright_fetch(query: str) -> str | None:
            with sync_playwright() as p:
                browser = p.chromium.launch(headless=True, args=["--no-sandbox"])
                page = browser.new_page()
                page.goto("https://www.ti.com/packaging/docs/partlookup.tsp", timeout=60000)
                # search box
                for sel in ['input[placeholder="Search phrase"]', "input[type='text']"]:
                    try:
                        page.wait_for_selector(sel, timeout=5000)
                        page.fill(sel, query)
                        page.keyboard.press("Enter")
                        break
                    except Exception:
                        pass
                page.wait_for_timeout(2000)
                base = to_ti_base_slug(query)
                loc = page.locator(f"a:has-text('{base}')").first
                if loc.count() == 0:
                    loc = page.locator("a").first
                try:
                    with page.expect_navigation(timeout=30000):
                        loc.click()
                except Exception:
                    pass
                # find datasheet link
                pdf_url = None
                for sel in ["a:has-text('Datasheet')", "a[href$='.pdf']"]:
                    if page.locator(sel).count() > 0:
                        href = page.locator(sel).first.get_attribute("href")
                        if href and href.lower().endswith(".pdf"):
                            pdf_url = requests.compat.urljoin(page.url, href); break
                if not pdf_url:
                    for a in page.locator("a[href]").all():
                        href = a.get_attribute("href") or ""
                        if href.lower().endswith(".pdf") and ("/lit/ds/" in href.lower() or "datasheet" in (a.inner_text() or "").lower()):
                            pdf_url = requests.compat.urljoin(page.url, href); break
                browser.close()
                return pdf_url

        for q in candidates[:3]:
            pdf_url = playwright_fetch(q)
            if pdf_url:
                pdf_path = download_pdf(pdf_url)
                manifest.update({
                    "ok": True, "method": "playwright",
                    "picked_query": q, "base_slug": to_ti_base_slug(q),
                    "product_url": ti_product_url(to_ti_base_slug(q)),
                    "pdf_url": pdf_url, "pdf_path": pdf_path
                })
                break
    except Exception as e:
        print("Playwright fallback error:", e)

# -------- save + report --------
man_path = os.path.join(OUT_DIR, "fetch_manifest.json")
with open(man_path, "w") as f:
    json.dump(manifest, f, indent=2)

if manifest["ok"]:
    print("\n✅ Datasheet fetched")
    print("Method      :", manifest["method"])
    print("Picked query:", manifest["picked_query"])
    print("Base slug   :", manifest["base_slug"])
    print("Product URL :", manifest["product_url"])
    print("PDF URL     :", manifest["pdf_url"])
    print("Saved PDF   :", manifest["pdf_path"])
    print("Manifest    :", man_path)
else:
    print("\n❌ No datasheet fetched.")
    print("Tried:", manifest["tried"][:5])
    print("Manifest:", man_path)
```
